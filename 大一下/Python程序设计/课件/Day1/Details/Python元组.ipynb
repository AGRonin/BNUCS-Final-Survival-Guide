# Python元组Tuple

- 外形：( )
- 元素：可以不同类型
- 特点：序列，不可变

<font color='red'><b> 元组适用于保存不修改，只读取的序列数据。</b></font>

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
#创建元组
t = (1)
print(t, type(t))

t = 1,   #逗号对元组很重要
print(t, type(t))

t = (1,)
print(t, type(t))

print( (), tuple())  #空元组

t1 = (5, 4, 1, 6)
t2 = tuple('apple')
print(t1, t2)
```

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
#容器/序列的通用操作
for i in t1:
    print(i)

print(len(t1), len(t2))
print(max(t1), min(t2), sum(t1))
print(sorted(t1))
print(t1+t2, t1*3)

print(t2.count('p'))
print(t1.index(6))

for i,j in enumerate(t2, 1):
    print(i,j)
    
for i in zip(t1,t2):
    print(i)
    
for i in reversed(t1):
    print(i, end=' ')
```
