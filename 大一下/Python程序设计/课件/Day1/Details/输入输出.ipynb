# Python的输入和输出

## 1. python的输入

input()：接收键盘输入，无论输入的是数字、字母还是中文字符等，回车后返回的都是一个字符串。

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
a = input('请输入一个整数：')

#查看它的类型
print(type(a))

#修改类型
a = int(a)
#再次查看类型
type(a)
```

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    轮到你了：
</div>

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    更多代码示例：
</div>

```python
name = input('请输入要买的水果：')
print(name,type(name))

price = input('请输入单价：')
print(price,type(price))
price = float(price)
print(price,type(price))
```
