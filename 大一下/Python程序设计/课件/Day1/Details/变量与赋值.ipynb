# 变量与赋值

## 1. 变量

变量在程序中创建的名字，用来表示程序中的某些“物体”，这个物体可以是一个数值、一组数据、一个文件、一张图像、一个程序等。

在Python中，变量大小写敏感，并需满足以下条件：
- 首字符是字母或下划线，其余可以是字母、下划线、数字
- 不能是关键字、预定义标识符

````python
'''查看Python中的关键字'''
import keyword

print('Python关键字个数为：',len(keyword.kwlist))
print(keyword.kwlist)
````

````python
'''查看预定义标识符'''
print(dir(__builtin__))
````

<font color='red'>小技巧：</font>一般的Python开发环境（IDE）都有颜色提示，变量如果不是黑色，而是别的颜色，则很可能用了预定义的标识符或者关键字，需要重新取名。

## 2. 变量赋值

通过“=”给变量赋值，与C语言不同，Python是动态的强类型语言，Python中变量赋值时同时获得类型和值，即变量的类型不需要显示声明，而是根据值确定类型，并通过“引用”把变量和值连接起来。

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
a = 1.5
print(type(a),id(a), a) #type函数查看类型，id函数查看值的内存地址

b = 6
print(type(b),id(b), b)

c = 'hello'
print(type(c), id(c), c)

d = ['a', 1, 2.2]
print(type(d),id(d), d)

e = {1:'张三', 2:'李四'}
print(type(e), id(e), e)
```

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    更多赋值代码示例：
</div>

```python
'''多重赋值'''
a = b = 3.5
print(a,b, type(a), type(b), id(a), id(b))


'''多元赋值'''
a,b,c = 1,2,3 
print(a,b,c,type(a), type(b), type(c),id(a), id(b),id(c))

t=['a',1.5, 10]
a,b,c = t
print(a,b,c,type(a), type(b), type(c),id(a), id(b),id(c))


'''多元赋值还可以用于交换'''
x=1; y=2  #多条语句写在一行，用分号隔开
print(x,y,type(x), type(y), id(x), id(y))

x,y=y,x
print(x,y,type(x), type(y), id(x), id(y))


'''自变赋值（增量赋值）:+=  -=  *=  /=  %=  **=   <<=   >>=   &=   ^=  |='''
x /=5  #即x = x/5
y **=2 #即 y = y**2
print(x, y, type(x), type(y), id(x), id(y) )

```
