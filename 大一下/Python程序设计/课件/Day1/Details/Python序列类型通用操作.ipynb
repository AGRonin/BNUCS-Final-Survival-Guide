</div>

```python
#查看帮助文档
reversed?

a = [5, 7, 1.0, 1, True, 5.5]

b = reversed(a)
print(b, a)   #懒惰

for i in b:
    print(i, end=',')
    
print('\n***********')

#会消耗，只能访问一次，第二次遍历为空
for i in b:
    print(i, end=',') 

#可以转换成list，tuple等
c = list(reversed(a))
print(c, '-----', type(c))
```

####  enumerate

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
#查看帮助文档
enumerate?

a=['apple', 'banana','pear']

b = enumerate(a) 
print(b)  #懒惰
for i in b:
    print(i)

print('**********')
#同reversed 一样，会消耗
for i in b:
    print(i)

#默认参数start为0，修改一下，例如让它为2
b = enumerate(a, 2)
for i,j in b:
    print(i, j)
    
#可以转换成list，tuple等
c = tuple(enumerate(a))
print(c, '-----', type(c))
```

####  zip

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
#查看帮助文档
zip?

a=['apple', 'banana','pear']
b='123'

c = zip(a,b)
print(c) #懒惰
  
for i in c:
    print(i)

print('**********')
#同reversed 一样，会消耗
for i in c:
    print(i)

print('**********')
#可以2个以上对象进行拼接
c = zip(a, b, ([1,2], 0.1, 0.5))
for i in c:
    print(i)

print('**********')
#可以不同长度的对象进行拼接
for i in zip(a, 'abcd'):
    print(i)
```

### 2.7 序列拆封操作

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
a = [[1,2], 'a','b']

#相当于多元赋值，注意左边的变量的数量要和右边值的数量一致
x,y,z = a
print(x, '\n', y, '\n', z)

(x1,x2),y,z = a
print(x1, '\n', x2, '\n', y, '\n', z)

#*变量可以接收多个值，这些值打包成一个元组赋值给*变量
x,*y,z = range(10)
print(x, '\n', y, '\n', z)

*x,y,z = range(10)
print(x, '\n', y, '\n', z)

x,y,*z = range(10)
print(x, '\n', y, '\n', z)

#临时变量，也就是没有名字的变量
_,x,_,_ = range(4)
print(x)

#临时变量也可以加*，接收多个变量
*_,x,_ = range(4)
print(x)
```
