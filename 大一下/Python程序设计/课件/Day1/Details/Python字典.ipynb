### 1.字典的创建和定义

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
#创建一个空字典
d1 = {}
d2 = dict()
print(d1,d2)

#创建时指定key-value
d3 = {1:'apple', 2:'banana'}
print(d3)

#把别的类型转换成dict，有要求：每项元素长度为2
d4 = dict([['name', 'Tom'],('ID', '2021111')])
print(d4)

#关键字创建字典
d5 = dict(name = 'Tom', age = 19)
print(d5)

#复制
d6 = d5.copy()
print(d6, id(d6),id(d5))

#已知键值，创建字典
lst=['name', 'ID', 'age']
d7 = {}.fromkeys(lst)
print(d7)

d8 = dict.fromkeys(lst) #同上
print(d8)  

d9 = dict.fromkeys(lst, '缺失') #设置默认value
print(d9)  
```

### 3. 通过key查看value

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
d = {'apple': 5, 'banana':10}
print(d['apple'])
print(d['pear']) #出错

print(d.get('apple'))
print(d.get('pear')) #不出错，返回None
print(d.get('pear', '缺失')) #不出错，返回设定值

#setdefault有双重作用，key存在时和get类似，key不存在时增加key:value对
print(d.setdefault('banana'))

print(d.setdefault('pear')) #'pear'不存在，则添加，默认value为None
print(d)
```

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    轮到你了：
</div>

### 8. 字典的删除元素操作

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
#clear清空所有项
d = {'apple': 5, 'banana':10}
d.clear()
print(d)

#通过del删除
d = {'apple': 5, 'banana':10}
del d['apple']
print(d)

del d[1] #出错，1不是d的key

#通过pop删除
d = {'apple': 5, 'banana':10}
print(d.pop('apple'))
print(d)

d.pop(1)#同del一样，出错

print(d.pop(1, '缺失')) #不出错
print(d)


#popitem不带参数，弹出最后一项
d = {'apple': 5, 'banana':10}
print(d.popitem())
print(d)
print(d.popitem())
print(d)
print(d.popitem()) #出错，d已经空了
```
