# Python数值类型

## 1. 整型
支持的长度与内存大小有关，可以处理很大很大的整数，基本无溢出问题。

整型有十进制、八进制（前缀0o/0O）、十六进制（前缀0x/0X）、二进制（前缀0b/0B）

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
a = 999999999999999999999999999999999999999999999999999999999999999999999999999
print(type(a),id(a), a) 

print(0b10, type(0b10), 0o10, type(0o10), 0x10, type(0x10))

'''不同进制转换: bin/oct/hex/int'''
print( bin(10) )  #十进制 -> 二进制
print( int(0b1010) )   #二进制 -> 十进制
print( oct(0b1010) )   #二进制 -> 八进制
print( hex(0b1010) )   #二进制 -> 十六进制

print( oct(10) )   #十进制 -> 八进制
print( int(0o12) )   # 八进制-> 十进制

print( hex(10) )   #十进制 -> 十六进制
print( int(0xa) )   # 十六进制-> 十进制
```

</div>

## 2. 布尔型

整型的子类：只有两个值： True、False

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
x = True
y = False
print(x, type(x), y, type(y))
print(int(x), int(y))
```

## 3. 浮点型

即数学中的实数，可以用科学记数法表示，有精度限制。

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
#精度限制，不确定尾数
a = 115.88888888888888888888888888888888
print(a, type(a))

b = 2.2222222222222222222222222222222222222222
print(b, type(b))

#科学记数法
print(-3.4e2, type(-3.4e2))

#精度限制
print( 1.1+2.2 == 3.3)

#正确判断操作
print( (1.1+2.2-3.3)<1e-6)
```

## 4. 复数型

实部+虚部j/J

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
x = 2.3+1.5j
print(x, type(x), 3.3J, type(3.3J), 1.8+0J, type(1.8+0j))

print(x.real, x.imag)

print(x.conjugate())
```

## 5. 数值类型转换

int -->float -->complex (往复杂的转)

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
#自动转
result = 1 + 4.5 
print(result, type(result))

result += (2+3j)
print(result, type(result))

#手动转
#把字符串转换成整型
print(int('1'))

#把浮点型转换成整型
print(int(2.3))

#把字符串转换成浮点型
print(float('3.14'))

#把整型转换成浮点型
print(int(float(10)))

#把字符串转换成复数型
print(complex('5.6+2.2j'))

#把整型转换成复数型
print(complex(1))

#把浮点型转换成复数型
print(complex(7.7))
```
