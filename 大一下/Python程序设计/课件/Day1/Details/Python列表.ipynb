## 1. 定义、容器/序列通用操作

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
l1 = [5, 4, 1, 6]
print(l1)

print(list(), []) #空列表

l2 = list('apple')

#序列的通用操作
for i in l1:
    print(i)

a = sorted(l2)
print(a, l2)

print(l1+l2)
print(l2*2)

print(len(l1), len(l2))
print(min(l2), max(l2), sum(l1))

print(l2.count('p'))
print(l1.index(6))

for i,j in enumerate(l2, 1):
    print(i,j)
    
for i in zip(l1,l2):
    print(i)
    
for i in reversed(l1):
    print(i, end=' ')
```

## 4. 列表解析式

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
l1 = [i**2 for i in range(5)]
print(l1)

l2 = [i**2 for i in range(5) if i%2==1]
print(l2)

l3 = [(i+j, i*j) for i in [10, 11] for j in [1,2,3]]
print(l3)
```

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    轮到你了：
</div>

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
#解析式生成字典
d = {i[0]: i[1] for i in zip([1,2,3], ['a','b','c'])}
print(d)

#解析式生成集合
a = [1,2,5,2,8,1,4]
s = {i for i in a}
print(s)
print(set(a))

s2 = {i for i in a if i%2==0}
print(s2)
```

###  <span style ="color: red">遍历的同时修改list 安全策略 </span>

![遍历的同时修改list](遍历的同时修改列表.png)

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
#删除列表中的所有1的正确姿势：
l = [1,1,2]
l2=[]
for i in l:
    if i!=1:
        l2.append(i)
print(l2) 

#别忘了还有列表解析式这个神器
l2 = [i for i in l if i !=1]
print(l2)
```
