### 3. set和frozenset的基本操作

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
#比较运算( ==, !=,  <, <=,  >, >= )
s1 = {1,2,3,4}

print(s1^s4)
```

### 4. set和frozenset的通用方法

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
#复制
s1 = {1,2,3,4}
s = s1.copy()
print(s,s1,id(s), id(s1))


#判断两个集合是否有交集，一个是否是另一个的子集或超集等
s1 = {1,2,3,4}
s2 = {2,3}
print(s1.isdisjoint(s2)) #如果s1和s2交集为空，返回True，否则False
print(s1.issubset(s2)) #如果s1是s2的子集，返回True，否则False
print(s1.issuperset(s2)) #如果s1是s2的超集，返回True，否则False

#用方法实现 交、并、差、对称差运算
s3 = {3,4,5,6}
s = s1.intersection(s3)
print(s, s1) #s1不变

s = s1.union(s3)
print(s, s1) #s1不变

s = s1.difference(s3)
print(s, s1) #s1不变

s = s3.difference(s1)
print(s, s3) #s3不变

s = s1.symmetric_difference(s3)
print(s, s1) #s1不变

#多个集合的交、并、差运算
print(s1.intersection(s3, s2))
print(s1.union(s3, s2))
print(s1.difference(s3, s2))
```

### 5. set的独有操作

#### 5.1 会改变自己的交、并、差、对称差运算

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
s1 = {1,2,3,4}
s2 = {2,3}
s3 = {3,4,5,6}

s1.update(s2,s3)  #相当于s1 = s1|s2|s3
print(s1)

s1 = {1,2,3,4}
s1.intersection_update(s2,s3)  #相当于s1 = s1&s2&s3
print(s1)

s1 = {1,2,3,4}
s1.difference_update(s2,s3)  #相当于s1 = s1-s2-s3
print(s1)

s1 = {1,2,3,4}
s1.symmetric_difference_update(s2)  #相当于s1 = s1^s2
print(s1)
```
