### 3.1 Python程序的组成及风格

```python
"""
Created on Tue Mar  5 09:04:16 2024

@author: DengQQ
"""

#功能：根据用户输入的花瓣数量和颜色绘制一朵彩色花

import turtle  #导入绘图模块

n = int(input("请输入花瓣的数量："))  #input接收键盘输入，返回字符串
colors = input("请输入花瓣的颜色，各颜色间逗号隔开：").split(",")

print(n, colors)

colorsNum = len(colors)  #颜色的数量

turtle.pensize(5)

for i in range(int(n)):
    if i==n-1 and n%colorsNum==1:  #如果花瓣数量是颜色数量的整数倍+1，并且当前画最后一片花瓣
        turtle.pencolor(colors[1])
    else:
        turtle.pencolor(colors[i%colorsNum])
    turtle.circle(100) #画圆形花瓣
    turtle.left(360/n)  #向左转一个角度
    
turtle.done()
```

##### 循环结构程序示例1：

利用random库，随机出100道两个数的加/减口算题。要求：

(1)两个运算数以及得数的范围均为[0,99]；

(2)如果是加法运算，要求有进位；如果是减法运算，要求有退位；

(3)在屏幕上打印出结果，每一行显示4道题，注意对齐。

```python
import random

for i in range(1,101):  #C语言：for (int i=1; i<101; i++)
    satisfied = False
    while not satisfied:
        a= random.randint(0,99)
        b= random.randint(0,99)
        c=random.choice(['+','-'])
        satisfied = ((c=='+') and (a+b<100) and (a%10+b%10>10) \
            or(c=='-') and (a-b>0) and (a%10<b%10))
            
    print(f'{a:2}{c}{b:2}=', end='\t\t' if i%4!=0 else '\n')
```

##### 循环结构程序示例2：循环嵌套、else子句示例（2-100的素数合数判断）

```python
primes = []  #初始化为空列表
composites = []

for i in range(2,101):
    for j in range(2, i):
        if i%j == 0:
            composites.append(i)
            break #可以整除，是合数，结束对该数字的判断
    else:
        primes.append(i)
    
print(f"2-100中合数有：{composites}")
print(f"2-100中素数有：{primes}")
```

```python
print(f"2-100中合数有：", end='')
for i in composites:  #i依次取列表中的元素
    print(i, end=' ')
print()

print(f"2-100中素数有：", end='')
for i in primes:
    print(i, end=' ')
print()
```

## 5. turtle绘图

##### 分支循环结构+turtle程序示例：用蒙特卡罗方法计算圆周率

<img src = "Images\计算圆周率.png" alt="计算圆周率"  width="65%" height="65%"/>

```python
from random import random

dots=100*100  #总沙子数
ins=0

for i in range(dots):
    x,y=random(),random()  #随机撒一粒沙子
    dis=pow(x**2+y**2,0.5)  #计算到圆心的距离

    if dis<=1:
        ins+=1  #在圆里的沙子数量增加1
    
pi = 4*ins/dots

print("圆周率pi={}".format(pi))    
```

```python
#利用turtle库对上述过程进行可视化
from turtle import *

dots=100*100
ins=0
speed(0)  #最快龟速
tracer(1000)  #每1000个动作更新一下屏幕
pensize(2)
for i in range(dots):
    x,y=random(),random()
    dis=pow(x**2+y**2,0.5)
    if dis<=1:
        ins+=1
        pencolor('red')  #圆里的颜色为红色
    else:
        pencolor('blue')  #圆外的颜色为蓝色
        
    penup()
    goto(400*x,400*y)
    pendown()
    dot()   #画一个点
    
pi = 4*ins/dots
penup()
goto(150,-50)
pendown()
pencolor('black')
write("圆周率pi={}".format(pi), align="left",\   
             font=("Arial", 20, "bold"))    
done()
```

### 6.1 容器类型的通用操作

#### (1) for循环遍历元素

容器都是可迭代的，可以通过for循环一一访问容器中的元素。

注意：对于集合的遍历，元素被访问的次序不确定，但都能访问到；对于字典的遍历，则只对key进行访问，value不参与。

```python
for c in 'a b*':  #字符串
    print(c)
    
print('-------------------------')

for i in (1,2,3):  #元组
    print(i)

print('-------------------------')
    
for i in [(1,2,3), 'a']:  #列表
    print(i)
```

```python
for i in {1, 'a', 3.5}:  #集合元素的遍历，顺序不定
    print(i)
```

```python
d = {1002:['李四', 'ls@abc.com', '13812341234'], 1001:['张三', 'sz@abc.com', '58801234', '13912345678']}
for i in d: #字典，for只对key遍历，不管value
    print(i)
```

#### 子任务2：进一步对李白诗歌内容进行字频统计，即统计每个字出现的次数，并对字频进行归一化处理（即除以诗歌篇数），然后输出显示字频统计结果，输出显示时根据字频降序排列，只打印频率大于等于0.1的结果；

```python
#字频统计

poems_num_li = len(poems_li)
print(poems_num_li)

#方法1，用count
wordsDct_li={}
for w in uniqwords_li:
    wordsDct_li[w] = words_li.count(w)/poems_num_li

print(wordsDct_li)
```

```python
#方法2   
wordsDct_li=dict.fromkeys(uniqwords_li,0)
print(wordsDct_li)
```

```python
for w in words_li:
    wordsDct_li[w]+=1/poems_num_li

print(wordsDct_li)
```

```python
#方法3，用标准库求词频    
from collections import Counter
wordsDct_li = Counter(words_li)

print(type(wordsDct_li))
print(wordsDct_li)
```

```python
#归一化
for w in wordsDct_li:
    wordsDct_li[w] /= poems_num_li
    
print(wordsDct_li)
```

```python
'''根据词频逆序排序'''

#方法1，直接对字典排序，设置排序规则：根据key对应的value进行排序
sort_words_li=sorted(wordsDct_li, key=wordsDct_li.get, reverse=True)  

#排序后只有key，没有value
print(sort_words_li)
```

```python
print('汉字---频率')
for w in sort_words_li:
    if wordsDct_li[w]<0.1:
        break
    print(f'{w}   {wordsDct_li[w]:10.8f}')
```

```python
#方法2：可以自己建一个列表，列表中每一项元素为（value，key）

sort_words_li = [(wordsDct_li[i],i) for i in wordsDct_li]
sort_words_li
```

```python
sort_words_li.sort(reverse = True)
sort_words_li
```

```python
print('汉字---频率')
for x,y in sort_words_li:
    if x<0.1:
        break
    print(f'{y}   {x:10.8f}')
```

#### 子任务3：采用同上述相同的方法，得到杜甫诗歌的字频统计结果。在此基础上获得李白和杜甫的高频词集合（即频率大于等于0.1的汉字集合），获得两位诗人都爱用的字、李白爱用但杜甫不爱用的字以及杜甫爱用而李白不爱用的字，并输出对应结果。

```python
# 读入杜甫.txt，用上述同样的方法得到杜甫的字频

#读文件    
with open('files/杜甫.txt', encoding='utf-8') as f:
    poems_text_du=f.read() 

#按行分割
poems_du=poems_text_du.splitlines()

#获得诗歌内容
contents_du=[one.split()[-1] for one in poems_du]

#获得诗词中的文字列表
words_du=[i for one in contents_du for i in one if i.isalpha()]

#去重
uniqwords_du=set(words_du)

#求词频
poems_num_du = len(poems_du)
wordsDct_du=dict.fromkeys(uniqwords_du,0)
for w in words_du:
    wordsDct_du[w]+=1/poems_num_du
```

```python
'''两位诗人的比较'''

#分别获得两个的高频词，结果为两个集合
li_like_words={i for i in wordsDct_li if wordsDct_li[i]>=0.1}
du_like_words={i for i in wordsDct_du if wordsDct_du[i]>=0.1}

print(li_like_words)
print('---------------------------------------------------------')
print(du_like_words)
```

```python
#集合操作
common_words=li_like_words & du_like_words   #李白和杜甫都爱用的
only_li_words=li_like_words - du_like_words  #李白爱用杜甫不爱用
only_du_words=du_like_words-li_like_words    #杜甫爱用李白不爱用
```

```python
print('两位诗人都爱用的字有{}个，这些字为：'.format(len(common_words)))
for count,i in enumerate(common_words, 1):
    print('{:3}'.format(i), end = '\n' if count%10==0 else '')
print()
```

```python
print('李白爱用但杜甫不爱用的字有{}个，这些字为：'.format(len(only_li_words)))
for count,i in enumerate(only_li_words, 1):
    print('{:3}'.format(i), end = '\n' if count%10==0 else '')
print()
```

```python
print('杜甫爱用但李白不爱用的字有{}个，这些字为：'.format(len(only_du_words)))
for count,i in enumerate(only_du_words, 1):
    print('{:3}'.format(i), end = '\n' if count%10==0 else '')
print()
```

```python
"""利用背景图片生成词云,设置停用词词集"""
import numpy as np
from PIL import Image
from wordcloud import STOPWORDS

text = open('Files/alice.txt').read()

# read the mask image
alice_mask = np.array(Image.open("Files/alice_mask.png"))   
# alice_mask = np.array(Image.open("Files/alice_color.png"))  #mask可以不是黑白图像

stopwords = set(STOPWORDS)  #停用词
stopwords.add("said")  #在集合STOPWORDS的基础上增加词

# Generate a word cloud image
wc = WordCloud(background_color="pink", 
               max_words=2000, 
               mask=alice_mask,
               stopwords=stopwords,  #不参与词云的词语
               contour_width=3, contour_color='black')  #设定图像有边界，边界的宽度和颜色
wc.generate(text)

# show
plt.imshow(wc, interpolation='bilinear')
plt.axis("off")
plt.show()
```

### 7.4 中文分词

jieba官网：https://github.com/fxsjy/jiebaist


**分词函数：jieba.cut (sentence, cut_all=False, HMM=True, use_paddle=False)**
- sentence：需要分词的字符串
- cut_all：是否全模式，默认为否，采用的是精确模式
- HMM：是否使用 HMM(Hidden Markov Model)模型
- use_paddle：是否使用paddle模式

返回一个可迭代的 generator。

```python
import jieba
seg1 = jieba.cut("我来到北京师范大学", cut_all = True)  #全模式
seg2 = jieba.cut("我来到北京师范大学", cut_all = False) #精确模式

print(seg1, seg2)
print('全模式：' + '/'.join(seg1))
print('精确模式：' + '/'.join(seg2))
```

```python
#分词的同时获得词性
import jieba.posseg as pseg

s='我来到北京师范大学。'
words = pseg.cut(s) #jieba默认模式
for word,flag in words:
    print(word,flag)
```

```python
"""中文文本分词后生成词云"""

text = ' '.join(jieba.cut(poems_text_li))  #分词后重新组合

# Generate a word cloud image
wc = WordCloud(font_path='C:\Windows\Fonts\simsun.ttc', min_word_length=2)
wc.generate(text)

#show
plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.show()
```
