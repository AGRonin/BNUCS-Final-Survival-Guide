# 字符串

### 1. 序列通用操作

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
s = 'abcdefgabc'
print(s, type(s))

#索引、切片操作
print(s[2], s[1::2])

s2 = 'hijkl'
print(s+s2, s*3)

for i in s:
    print(i)

print(len(s), len(s2))
print(max(s), min(s2))
print(sorted(s))


print(s.count('a'), s.count('a', 1, 5))
print(s.index('c'), s.index('c', 3))

for i,j in enumerate(s, 1):
    print(i,j)
    
for i in zip(s,s2):
    print(i)
    
for i in reversed(s):
    print(i, end=' ')
```

### 2.1 类型判断

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
s = 'abc123'
print(s.isalnum())
print(('abc123.,').isalnum())

print('abc中文'.isalpha()) #中文也是alpha


#对数字的宽容程度：isdecimal<isdigit<isnumeric
print('123'.isdecimal(), '⑴'.isdecimal(), '十'.isdecimal())#这里的⑴是一个字符，不是括号里面加1
print('123'.isdigit(), '⑴'.isdigit(), '十'.isdigit()) 
print('123'.isnumeric(), '⑴'.isnumeric(), '十'.isnumeric()) 

print('I Love Python'.istitle())

print('abc'.islower())
print('ABC'.isupper())

print('ab9* !&\'/'.isprintable())
print('ab9* !&\'/\t\a'.isprintable())

print(' \n\t'.isspace())
```

### 2.5 拆分和组合

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
print('123 abc 123\nabc\t123'.split())

print('123 abc 123\nabc\t123'.split(' '))

print('123 abc 123\nabc\t123'.split(' ', 1))

print('123 abc 123\nabc\t123'.rsplit(maxsplit=2))

x,y,z = '12,3,5'.split(',')
print(x,y,z)

print('12,3,5'.partition(','))
print('12,3,5'.rpartition(','))

print(','.join(['12','3','5']))

print(''.join(['12','3','5']))
```

### 2.6 编码转换

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
print('你好!'.encode()) #utf-8编码
print('你好!'.encode('unicode-escape')) #unicode编码
print('你好!'.encode('cp936')) #cp936编码

print(b'\xe4\xbd\xa0\xe5\xa5\xbd!'.decode()) #utf-8编码
print(b'\xe4\xbd\xa0\xe5\xa5\xbd!'.decode('unicode-escape')) #乱码或错误
print(b'\xe4\xbd\xa0\xe5\xa5\xbd!'.decode('cp936')) #乱码或错误

print(b'\\u4f60\\u597d!'.decode('unicode-escape')) #unicode编码
print(b'\xc4\xe3\xba\xc3!'.decode('cp936')) #cp936编码
```
