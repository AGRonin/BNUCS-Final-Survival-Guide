计算出租车费，计价规则如下：

1.如果目的地是市内区域，则起步价为8元，起步距离为2公里，超过2公里的部分，每1公里收2元；

2.如果目的地是市外区域，则起步价为15元，起步距离为5公里，超过5公里的部分，每1公里收3元。

编写程序，用户输入目的地是否在市内，以及路程长度，输出打车费用。

```python
dis = int(input('请输入路程长度：'))
in_city = int(input('目的地是否在市区，在--1，不在--0：'))
if in_city==1:
    if dis<=2:
        print('打车费为8元！')
    else:
        money = 8+ (dis-2)*2
        print('打车费为{}元！'.format(money))
else:
    if dis<=5:
        print('打车费为15元！')
    else:
        money = 15+ (dis-5)*3
        print('打车费为{}元！'.format(money))
```

```python

names = "张无忌,黄蓉"
for i in names:
    if i == ',' :  #如果是逗号，则终止循环，这里表现为打印出张无忌后跳出循环
        break
    print(i,end="")
print("\n执行循环体外的代码")  


while True:
    b = input('任意输入：')
    if b == '1':  #输入为1时终止循环
        break
    print('在while循环中...')
print('结束')
```

```python
#比较下面两组代码的结果

for i in range(3):
    for j in range(3):
        if i == j == 1:
            break
        else:    
            print (i, '----', j)



for i in range(3):
    for j in range(3):
        if i == j == 1:
            break
        else:    
            print (i, '----', j)
    else:
        continue
    break  
```
