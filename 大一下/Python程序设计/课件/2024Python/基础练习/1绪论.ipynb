## 5. python的运算符和表达式

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
#算术运算
a = 12
b = 5
c = 3
print(b**3, a*b, a+b, a-b, a//b, a%b, a/b)

#比较运算
print(a==b, a>=b, a<b, a!=b)

#逻辑运算
print(not(a==b), (a==b) and (c>b), (a==b) or (a>b))

#位运算符：把数字看作二进制来进行计算，平时研究不涉及二进制的可以不用了解
a = 60   # 60的二进制为 00111100 
b = 13   # 13的二进制为 00001101 
 
print('十进制{0}，二进制{0:08b}'.format(a & b))     # 12 的二进制为 00001100

print('十进制{0}，二进制{0:08b}'.format(a | b)) # 61 = 00111101

print('十进制{0}，二进制{0:08b}'.format(a ^ b)) # 49 = 00110001

print('十进制{0}，二进制{0:08b}'.format(~a)) # -61 = 11000011

print('十进制{0}，二进制{0:08b}'.format(a << 2))  # 240 = 11110000

print('十进制{0}，二进制{0:08b}'.format(a >> 2))  # 15 = 00001111

#连写
print(c<b<a)
```
