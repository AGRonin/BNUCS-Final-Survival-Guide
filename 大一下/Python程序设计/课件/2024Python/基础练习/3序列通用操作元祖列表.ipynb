### 1.7 序列拆封操作

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
a = [[1,2], 'a','b']

#相当于多元赋值，注意左边的变量的数量要和右边值的数量一致
x,y,z = a
print(x, '\n', y, '\n', z)

(x1,x2),y,z = a
print(x1, '\n', x2, '\n', y, '\n', z)

#*变量可以接收多个值，这些值打包成一个元组赋值给*变量
x,*y,z = range(10)
print(x, '\n', y, '\n', z)

*x,y,z = range(10)
print(x, '\n', y, '\n', z)

x,y,*z = range(10)
print(x, '\n', y, '\n', z)

#临时变量，也就是没有名字的变量
_,x,_,_ = range(4)
print(x)

#临时变量也可以加*，接收多个变量
*_,x,_ = range(4)
print(x)
```

## 4. 列表解析式

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
l1 = [i**2 for i in range(5)]
print(l1)

l2 = [i**2 for i in range(5) if i%2==1]
print(l2)

l3 = [(i+j, i*j) for i in [10, 11] for j in [1,2,3]]
print(l3)
```

###  <span style ="color: red">遍历的同时修改list 安全策略 </span>

![遍历的同时修改list](遍历的同时修改列表.png)

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
#删除列表中的所有1的正确姿势：
l = [1,1,2]
l2=[]
for i in l:
    if i!=1:
        l2.append(i)
print(l2) 

#别忘了还有列表解析式这个神器
l2 = [i for i in l if i !=1]
print(l2)
```
