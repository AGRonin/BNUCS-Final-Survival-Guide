### 1.字典的创建和定义

<div class="alert alert-success alertsuccess" style="margin-top: 10px">

#已知键值，创建字典

### 4. set和frozenset的通用方法

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
#复制
s1 = {1,2,3,4}
s = s1.copy()
print(s,s1,id(s), id(s1))


#判断两个集合是否有交集，一个是否是另一个的子集或超集等
s1 = {1,2,3,4}
s2 = {2,3}
print(s1.isdisjoint(s2)) #如果s1和s2交集为空，返回True，否则False
print(s1.issubset(s2)) #如果s1是s2的子集，返回True，否则False
print(s1.issuperset(s2)) #如果s1是s2的超集，返回True，否则False

#用方法实现 交、并、差、对称差运算
s3 = {3,4,5,6}
s = s1.intersection(s3)
print(s, s1) #s1不变

s = s1.union(s3)
print(s, s1) #s1不变

s = s1.difference(s3)
print(s, s1) #s1不变

s = s3.difference(s1)
print(s, s3) #s3不变

s = s1.symmetric_difference(s3)
print(s, s1) #s1不变

#多个集合的交、并、差运算
print(s1.intersection(s3, s2))
print(s1.union(s3, s2))
print(s1.difference(s3, s2))
```

# 浅拷贝与深拷贝

## 1. 浅拷贝

<div class="alert alert-success alertsuccess" style="margin-top: 10px">
    代码示例：
</div>

```python
#浅拷贝
jack = ['Jack', ['age', 20]]
tom = jack.copy()

print(id(jack),id(jack[0]), id(jack[1]), id(jack[1][0]),id(jack[1][1]))
print(id(tom),id(tom[0]), id(tom[1]), id(tom[1][0]),id(tom[1][1]))

tom[0] = 'Tom'
print(jack, tom)
print(id(jack),id(jack[0]), id(jack[1]), id(jack[1][0]),id(jack[1][1]))
print(id(tom),id(tom[0]), id(tom[1]), id(tom[1][0]),id(tom[1][1]))

tom[1][1] = 25
print(jack, tom)
print(id(jack),id(jack[0]), id(jack[1]), id(jack[1][0]),id(jack[1][1]))
print(id(tom),id(tom[0]), id(tom[1]), id(tom[1][0]),id(tom[1][1]))
```
